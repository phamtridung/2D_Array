### Exercise 1: Reading a grade table
Grades of 4 students across 3 tests. Each row is a student and each column is a test.

import numpy as np
scores = np.array([[7, 8, 6],
                   [9, 5, 7],
                   [6, 9, 8],
                   [8, 7, 9]])

Objectives:
(a) Print the shape, ndim, and size of scores. Explain what the numbers 4 and 3 mean.
(b) Get the score of the 3rd student on the 2nd test (note that indices start at 0).
(c) Get all scores of the first test (one column).
(d) Get the scores of the first 2 students on the last 2 tests, using only one slicing statement.

```python
import numpy as np
scores = np.array([[7, 8, 6],
                    [9, 5, 7],
                    [6, 9, 8],
                    [8, 7, 9]])

print(f'shape: {scores.shape}')
# (4,3) represents the dimensions of the array. 4 represent the number of rows and 3 represent the number of collumns
print(f'ndim: {scores.ndim}')
print(f'size: {scores.size}')
print(f'scores of 3rd student, 2nd test: {scores[2, 0]}')
print(f'Scores of the first test: {scores[:, 0]}')
print(f'Score of first 2 students on the last 2 tests: {scores[0:2, 1:3]}')
```

### Exercise 2: Averages by row and by column
Reuse the array scores from Exercise 1.
(a) Compute the average score of each student (rounded to 2 decimals).
(b) Compute the average score of each test.
(c) Find the highest score of each test and the lowest score of each student.
(d) Predict first, run later: scores.sum(axis=0).shape and scores.sum(axis=1).shape — what are
they? Why?

```python
import numpy as np
scores = np.array([[7, 8, 6],
                    [9, 5, 7],
                    [6, 9, 8],
                    [8, 7, 9]])

student_average = scores.mean(axis=1)
for i in range(len(student_average)):
    avg = student_average[i]
    print(f'Student {i+1} average is: {round(avg, 2)}')

test_average = scores.mean(axis=0)
for k in range(len(test_average)):
    test_avg = test_average[k]
    print(f'Test {k+1} average is: {round(test_avg, 2)}')

test_max = scores.max(axis=0)
for j in range(len(test_max)):
    highest_score = test_max[j]
    print(f'Highest score of test {j+1} is {highest_score}')

student_min = scores.min(axis=1)
for l in range(len(student_min)):
    lowest_score = student_min[l]
    print(f'Lowest score of student {l+1} is: {lowest_score}')

#scores.sum(axis=0).shape --> (3, 0) because it measures along the collumn, scores.sum(axis=1).shape --> (0, 4) because it measures along the rows
```

### Exercise 3: A 6×6 image and masks
A small grayscale image (0 = black, 255 = white). The bright region is in the middle of the image.

img = np.array([[ 10, 20, 30, 40, 50, 60],
[ 15, 200, 210, 205, 55, 65],
[ 20, 215, 255, 220, 60, 70],
[ 25, 210, 225, 215, 65, 75],
[ 30, 35, 40, 45, 70, 80],
[ 35, 40, 45, 50, 75, 85]])

Objectives:
(a) Crop the 3×3 bright region in the middle (rows 1–3, columns 1–3) and print it.
(b) Create mask = img > 100. How many bright pixels are there?
(c) Copy the image, then turn it into a binary image: pixel ≤ 100 → 0, pixel > 100 → 255. The original image
must not change.
(d) What percentage of the image do the bright pixels make up? (hint: mask.mean() is the proportion of
True values)

```python
img = np.array([[ 10, 20, 30, 40, 50, 60],
                [ 15, 200, 210, 205, 55, 65],
                [ 20, 215, 255, 220, 60, 70],
                [ 25, 210, 225, 215, 65, 75],
                [ 30, 35, 40, 45, 70, 80],
                [ 35, 40, 45, 50, 75, 85]])

print("Cropped image of bright region: ")
print(img[1:4, 1:4])

mask = img > 100
bright = mask.sum()
print("")
print(f'Number of bright pixels: {bright}')

binary_image = np.where(img <101, 0 ,255).astype(np.uint8)
print("")
print("Binary image: ")
print(binary_image)

bright_pixels = (bright / img.size)*100
print("")
print(f'Percentage of bright pixels in the image: {bright_pixels}')

```

### Exercise 4: Analyzing the shot-angle table
The table shots has 8 shots, each with 3 measured angles (degrees): elbow, knee, shoulder. The array made tells
whether each shot went in. Simulated data for NumPy practice, not to be used to draw conclusions about
shooting technique.

import numpy as np
shots = np.array([[88, 118, 92],
                  [92, 121, 95],
                  [85, 112, 88],
                  [95, 125, 99],
                  [90, 119, 94],
                  [80, 108, 84],
                  [93, 123, 97],
                  [78, 106, 82]]) # columns: elbow, knee, shoulder (degrees)
made = np.array([True, True, False, True, True, False, True, False])

Objectives:
(a) Compute the average angle of all 8 shots for each angle type.
(b) Compute the average angles of the made shots and the missed shots (use the masks made and ~made).
(c) Compute the difference (made − missed) for each angle. In this data, which angle differs the most?
(d) Subtract each column’s mean from shots (bringing every column to mean 0). Verify with
np.allclose(..., 0).
(e) How many shots have an elbow angle > 90 and were made? (use &)

```python
import numpy as np

shots = np.array([[88, 118, 92],
                    [92, 121, 95],
                    [85, 112, 88],
                    [95, 125, 99],
                    [90, 119, 94],
                    [80, 108, 84],
                    [93, 123, 97],
                    [78, 106, 82],])

made = np.array([True, True, False, True, True, False, True, False])
columns = ['elbow', 'knee', 'shoulder']

average_angle = shots.mean(axis=0)
for i in range(len(average_angle)):
    if i == 0:
        print(f'Average elbow angle of 8 shots: {round(average_angle[0], 2)}')
    elif i == 1:
        print(f'Average knee angle of 8 shots: {round(average_angle[1], 2)}')
    elif i == 2:
        print(f'Average shoulder angle of 8 shots: {round(average_angle[2], 2)}')

made_shots = shots[made]
missed_shots = shots[~made]
average_made = made_shots.mean(axis=0)
average_miss = missed_shots.mean(axis=0)
print(f'Average made shots angle: {average_made.round(2)}')
print(f'Average miss shots angel: {average_miss.round(2)}')

difference = average_made - average_miss
max_diff = difference.max()
angle_type = 'a'
for j in range(len(difference)):
    if difference[j] == max_diff:
        angle_type = columns[j]

print(f'Angle {angle_type} differs the most: {max_diff.round(2)}')

new_array = shots - average_angle
new_mean = new_array.mean(axis=0)
print(np.allclose(new_mean, 0))

elbow_angle = shots[:, 0]
count = 0
for i in range(len(elbow_angle)):
    if elbow_angle[i] > 90 and made[i]:
        count += 1

print(f'Made elbow shots angles that are over 90 degrees: {count}')
```
