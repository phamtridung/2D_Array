```python
import numpy as np
scores = np.array([[7, 8, 6],
                    [9, 5, 7],
                    [6, 9, 8],
                    [8, 7, 9]])

print(f'shape: {scores.shape}')
# (4,3) represents the dimensions of the array. 4 represent the number of rows and 3 represent the number of collumns
print(f'ndim: {scores.ndim}')
print(f'size: {scores.size}')
print(f'scores of 3rd student, 2nd test: {scores[2, 0]}')
print(f'Scores of the first test: {scores[:, 0]}')
print(f'Score of first 2 students on the last 2 tests: {scores[0:2, 1:3]}')
```

```python
import numpy as np
scores = np.array([[7, 8, 6],
                    [9, 5, 7],
                    [6, 9, 8],
                    [8, 7, 9]])

student_average = scores.mean(axis=1)
for i in range(len(student_average)):
    avg = student_average[i]
    print(f'Student {i+1} average is: {round(avg, 2)}')

test_average = scores.mean(axis=0)
for k in range(len(test_average)):
    test_avg = test_average[k]
    print(f'Test {k+1} average is: {round(test_avg, 2)}')

test_max = scores.max(axis=0)
for j in range(len(test_max)):
    highest_score = test_max[j]
    print(f'Highest score of test {j+1} is {highest_score}')

student_min = scores.min(axis=1)
for l in range(len(student_min)):
    lowest_score = student_min[l]
    print(f'Lowest score of student {l+1} is: {lowest_score}')

#scores.sum(axis=0).shape --> (3, 0) because it measures along the collumn, scores.sum(axis=1).shape --> (0, 4) because it measures along the rows
```
